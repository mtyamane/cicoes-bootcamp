Recall that to create a list, you surround objects in that list with brackets

```python
myList = [12, 34, 16, 7, 2, 56]
print(myList)
```

To access sections of the list, you can use the colon operator

<LIST_NAME>[<START>:<STOP>]  this will return the objects from index <START> to <STOP> - 

```python
sectionOfList = myList[1:4]
print(sectionOfList)
```

Grids are created by making a list of lists

For plotting, the matplotlib.pyplot module needs to be imported:

import matplotlib.pyplot as plt

```python
import matplotlib.pyplot as plt
```

To plot a line graph, use the plot() function

With one list, it uses the indices as the x-axis

To show to the screen from a script, use the show() function

```python
plt.plot(numpyList)
plt.show()
```

If you want to specify the x values, it assumes when you put in two parameters that the first are all x-values and the second are all y-values

```python
xs = [3, 6, 8, 12, 16]
ys = [23, 64, 12, 53, 25]

print("xs =", xs)
print("ys =", ys)
```

```python
plt.plot(xs, ys)
plt.show()
```

To plot each point, not connected by a line, use the scatter() function

```python
plt.scatter(xs, ys)
plt.show()
```

Histograms show us the frequency that value appears in a list

Histograms can be plotted with the hist() function

```python
histList = [1,2,2,3,3,3,4,5,5,36]
plt.hist(histList, bins=35)
plt.show()
```

To give titles to the x-axis and y-axis, use xlabel() and ylabel()

```python
plt.hist(histList)
plt.xlabel("Cars")
plt.ylabel("Frequency")
plt.show()
```

```python
randomInts = np.random.randint(0, 10, 20)  # parameters: <START>, <STOP>, <NUM_IN_LIST>
plt.hist(randomInts)
plt.show()
```

```python
randomNormal = np.random.normal(170, 10, 250)  # parameter: <MEAN>, 1 std dev, <NUM_IN_LIST>
plt.hist(randomNormal)
plt.show()
```

To save the figures, use savefig() function which takes in the filepath as its parameter

```python
fpath = "figures/"
fname = "myFirstFigure.png"
plt.savefig(fpath+fname)
```

Because this a Notebook, the saved figure doesn't show anything!

All code in a Notebook to create a figure needs to be in the same cell as the save code

```python
randomNormal = np.random.normal(170, 10, 250)  # parameter: <MEAN>, 1 std dev, <NUM_IN_LIST>
plt.hist(randomNormal)
plt.show()

fpath = "figures/"
fname = "mySecondFigure.png"
plt.savefig(fpath+fname)
```

We have to take out show() when we save figures so that the figure is in the environment

(show takes it out of the environment to show the figure on your screen)

```python
randomNormal = np.random.normal(170, 10, 250)  # parameter: <MEAN>, 1 std dev, <NUM_IN_LIST>
plt.hist(randomNormal)

fpath = "figures/"
fname = "myThirdFigure.png"
plt.savefig(fpath+fname)
```
