Given the table below:

| Score | Grade |
| :--- | -----------: |
| >= 90 | A |
| >= 80 | B |
| >= 70 | C |
| >= 60 | D |
| < 60 | F |

create a function called 'scoreToGrade' that takes you score as input and returns your grade

```python
# write the function here
# (hint: should return a string, don't print from the function!)
def scoreToGrade(score):
    # takes a number score as input
    # returns a letter grade as a string
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    elif score >= 60:
        return "D"
    else:
        return "F"
```

```python
# test the function here
scor = 67
print("Grade should be D")
print("Grade from the function:", scoreToGrade(scor))
```

Create a function called 'lowestValue' that takes a list of positive numbers as input and returns the lowest lowest Value

Test with \[12, 3, 67, 42, 7]

```python
# write the function here
def lowestValue(lyst):
    # takes a list of positive numbers as input
    # returns the lowest number
    min = lyst[0]
    for value in lyst:
        if value < min:
            min = value
    return min
```

```python
# test the function here
testLyst = [12, 3, 67, 42, 7]
print("Lowest value should be 3")
print("Lowest value from function:", lowestValue(testLyst))
```

Copy and paste your quadratic code in the cell below and refactor to turn it into a function called 'quadratic' that takes in three parameters and returns two values

```python
# write the function here
import math

def quadratic(a, b, c):
    # performs the quadratic formula on three inputs
    # returns the value from addition and the value from subtraction, respectively
    # requires two variables to catch the output!

    inRoot = b**2 - 4*a*c

    # apply the quadratic formula
    x_add = (-b + math.sqrt(inRoot)) / (2*a)
    x_sub = (-b - math.sqrt(inRoot)) / (2*a)

    return x_add, x_sub
```

```python
# test the function here
x1, x2 = quadratic(2, 7, 3)
print("Should return -0.5 from adding")
print("Value returned from adding:", x1)
print()
print("Should return -3.0 from subtracting")
print("Value returned from subtracting:", x2)
```
